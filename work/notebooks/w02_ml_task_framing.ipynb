Success Metric: $F_1\text{-score}$ with Precision@KBenchmark: $F_1 \ge 0.75$. Because refresh candidates are imbalanced, raw accuracy is misleading. Precision prevents content teams from wasting effort on false alarms, while recall ensures critical traffic-dropping pages are not missed.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

Why ML Beats Fixed Rules: A simple if/else rule (e.g., drop > 20%) fails to capture non-linear patterns such as seasonal variations, competitor rank shifts, and search intent differences. Machine Learning learns from multiple correlated features simultaneously to identify genuine content decay.

```python

if 'traffic_change_pct' in df.columns:
    rule_flagged = (df['traffic_change_pct'] < -0.20).sum()
    print(f"Simple rule flagged {rule_flagged} out of {len(df)} rows.")
else:
    print(f"Total rows requiring multi-variable evaluation: {len(df)}")
```
