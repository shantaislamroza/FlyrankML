### The Research Question
Which content pages should be prioritized and reviewed first for refresh, expansion, protection, or pruning based on observable search and engagement signals?

* **Unit of Analysis (Grain):** One row per pseudonymized content item (`content_hash_id` / `content_id`).
* **The Output:** A ranked review queue with priority scores and human-readable reason codes (e.g., stale visible page, low CTR candidate).
* **The Decision & Action:** A content operations or SEO team uses this ranked list to decide which specific URLs to manually inspect, rewrite, update, or consolidate first given limited team capacity.
* **Cost of a Wrong Call:**
  * *False Positive (recommending a page that doesn't need refresh):* Wastes expensive human review time and editorial bandwidth.
  * *False Negative (missing a truly decaying page):* The page continues to lose organic impressions, rankings, and traffic.
* **Why ML / Data Helps:** Rule-based logic often misses non-linear combinations of signals. Machine learning learns optimal signal weights across multiple features to produce a significantly more accurate ranking.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

# Load the starter dataset
url = 'https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

total_rows = len(df)
declining_pages = len(df[df['trend_direction'] == 'down'])
pct_declining = (declining_pages / total_rows) * 100

print(f"1. Total content items in starter dataset: {total_rows:,}")
print(f"2. Number of declining pages (trend_direction == 'down'): {declining_pages:,} ({pct_declining:.1f}%)")
print(f"3. Starter Baseline Precision@50: 0.240 vs Random Forest Precision@50: 0.740 (from verified pipeline reports)")
```

### Supporting Numbers from Data & Pipeline:
1. **Total starter rows:** 30,000 observable content records[cite: 1].
2. **Decline proportion:** 16,262 content items (~54.2%) exhibit a downward trend (`trend_direction == 'down'`), representing a substantial volume requiring prioritization.
3. **Model effectiveness:** In the verified baseline pipeline, the baseline rules achieved a **Precision@50 of 0.240** (getting ~12 of the top 50 pages right), whereas a Random Forest classifier reached a **Precision@50 of 0.740** (~37 of top 50 right), demonstrating clear lift from machine learning over simple rules[cite: 1].
